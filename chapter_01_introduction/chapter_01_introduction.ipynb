**Svar:**

AI, ML och DL hänger ihop rent hierarkiskt. AI, artificiell intelligens, är det generella området och handlar om att skapa system som kan utföra uppgifter som kräver någon form av intelligens.

## Fråga 2

**Frågeställning:**

Vilka är de fyra problemkategorierna inom ML?

**Svar:**

De fyra problemkategorierna inom ML är:

- **Regression** – används för att prediktera numeriska värden, exempelvis priset på ett hus. (Oberoende variabel X och beroende variabel Y)
- **Klassificering** – används för att avgöra vilken kategori eller klass en observation (datapunkt)tillhör, exempelvis om ett mejl är spam eller inte eller om en kund kommer churna när man gör kundsegmentering.
- **Dimensionsreducering** – används för att minska antalet dimensioner eller variabler i ett dataset samtidigt som så mycket relevant information som möjligt bevaras.
- **Klustring** – används för att gruppera observationer med liknande egenskaper när det inte redan finns fördefinierade kolumner i datan.

En väldigt simplifierad tolkning jag har gjort för att hålla koll på dessa problemkategorier är följande:

- Regression = hur mycket?
- Klassificering = vilken klass?
- Dimensionsreducering = kan vi förenkla datan?
- Klustring = vilka hör ihop?

### a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?

**Svar:**

Syftet är att använda olika delar av datan för olika steg i modellutvecklingen.

- **Träningsdata** används för att träna modellen.
- **Valideringsdata** används för att jämföra och justera modellen.
- **Testdata** används sist för att se hur modellen fungerar på ny data.

Flödet blir alltså ungefär **train -> validation -> test**.

Testdatan ska hållas separat tills slutet, annars finns risken att man anpassar modellen efter den och får en missvisande utvärdering. Problem som kan uppstå under denna process är overfitting/underfitting.

### b) Vad är k-delad korsvalidering för något?

**Svar:**

K-delad korsvalidering innebär att träningsdatan delas upp i **k delar**. Vilket kan läsas som "antal delar".

Modellen tränas flera gånger. Varje gång används en del som valideringsdata och resten som träningsdata.

Till exempel vid **5-fold cross validation** (vilket vi använt oss av mycket som exempel) delas datan upp i 5 delar och modellen tränas 5 gånger.

Fördelen är att man får en mer stabil bild av hur bra modellen presterar än om man bara använder en enda uppdelning.

### c) Vad är root mean squared error (RMSE) för något?

**Svar:**

RMSE står för **Root Mean Squared Error** och används för att mäta hur stora fel en regressionsmodell gör. (Random Forest, Linear regression m.fl)

Man räknar ut skillnaden mellan de verkliga och predikterade värdena, kvadrerar felen, tar medelvärdet och därefter roten ur resultatet.

Ett lägre RMSE-värde betyder generellt att modellen gör bättre prediktioner.

### d) Vad är en hyperparameter för något? Vad är en parameter för något?

**Svar:**

En **parameter** är ett värde som modellen själv lär sig från träningsdatan.

En **hyperparameter** är ett värde som vi ställer in innan eller under träningen för att styra hur modellen ska lära sig. Detta ställer vi in med en gridsearch vilket tas upp i nästa fråga.

Exempel:
- Parameter: vikterna i en linjär regressionsmodell.
- Hyperparameter: exempelvis `max_depth` i ett beslutsträd.

### e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med
processen som genomförs? Läser vi dokumentationen: https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV så
ser vi att hyperparametern refit har standardvärdet True, vad innebär
det?

**Svar:**

**Grid search** används för att testa flera olika kombinationer av hyperparametrar för en modell.

Man skapar ett "grid" med olika värden, och sedan "searchar" modellen igenom dem olika kombinationerna för att hitta den som presterar bäst.

`refit=True` betyder att den bästa kombinationen används för att träna om modellen på hela träningsdatan efter sökningen.

### f) Vad är kategorisk data och hur hanteras det?

**Svar:**

**Kategorisk data** är data som består av olika kategorier istället för vanliga numeriska värden.

Det finns två vanliga typer:

- **Nominal data** – kategorier utan någon naturlig ordning, t.ex. färg eller land (röd, grön, blå följer ingen ordning).
- **Ordinal data** – kategorier med en tydlig ordning, ex. [låg, medel, hög], [löjtnant, kapten, major] osv.

För att ML-modeller ska kunna använda kategorisk data behöver den ofta kodas om till siffror/numeriska värden.

- **One-hot encoding** skapar en separat kolumn för varje kategori.
- **Dummy-variable encoding** liknar one-hot encoding, men en kategori tas bort för att undvika redundans. Kategorin som saknas är fortfarande med, men räknas in som att all data som inte är med faller under den borttagna kategorin.
- **Ordinal encoding** ger kategorier numeriska värden utifrån deras ordning. Exempelvis 1, 2, 3 till låg, medel, hög.

### g) Vad är feature engineering?

**Svar:**

**Feature engineering** innebär att man skapar variabler för att göra datan mer användbar för en ML-modell.

Det kan till exempel vara att skapa en ny variabel från befintliga kolumner, skala om värden eller omvandla data till ett format som modellen lättare kan lära sig från. Enkla exempel är födelsedatum -> ålder, längd + vikt = BMI

### h) Vad menas med principle of parsimony?

**Svar:**

**Principle of parsimony** innebär att man borde föredra en så enkel modell som möjligt, så länge den presterar tillräckligt bra.

En mer komplex modell är alltså inte automatiskt bättre, eftersom den kan bli svårare att tolka och ha större risk att överanpassa datan. En bra tumregel är att komplex/stora mängder data -> komplex modell. Enkel data -> enkel modell

### Vad menas med att en modell är en förenkling av verkligheten?

**Svar:**

En modell försöker efterlikna verkligheten, men kan bara ta upp de delar som är relevanta för problemet.

Den blir därför en förenklad representation av verkligheten och kommer aldrig kunna fånga alla faktorer som påverkar det problem man försöker lösa.

Ett exempel är en modell som ska förutsäga en persons lön. Den kanske använder utbildning, erfarenhet och ålder, men tar inte hänsyn till exakt allt som påverkar lönen i verkligheten.
